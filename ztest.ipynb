```python
url = "/home/tehe/Work/tfm/datasets/ines/executions/executions/03_preparewindowsds/v001/03_preparewindowsds_dataset.parquet"
df = pd.read_parquet(url)
print(df.head())
```

```python
#   OW_events PW_events
# 0        []      [63]
# 1        []      [63]
# 2        []      [63]
# 3        []      [63]
# 4        []      [63]
OVERVACION = "OW_events"
PREDICCION = "PW_events"
# Quiero que me hagas un buscador de compoentes, que por ejemmplo si pogno en src="10" y dst="20", que me muesrte las lineas donde tenga este contenido en los arrays.
def search_components(df, src, dst):
    filtered_df = df.copy()
    
    # Filtrar por src si no está vacío
    if src and src.strip():
        src_int = int(src)
        filtered_df = filtered_df[filtered_df[OVERVACION].apply(lambda x: src_int in x)]
    
    # Filtrar por dst si no está vacío
    if dst and dst.strip():
        dst_int = int(dst)
        filtered_df = filtered_df[filtered_df[PREDICCION].apply(lambda x: dst_int in x)]
    
    return filtered_df
```

```python
SRC="65"
DST="81"   

result = search_components(df, SRC, DST)
print(f"Resultados para src='{SRC}' y dst='{DST}':")
print(f"Total de líneas encontradas: {len(result)}")
print(result.head(10))
```

```python
import pandas as pd
# Asumo que df_total ya está definido en tu código anterior

separator = "=" * 80
# Configurar pandas para mostrar el contenido completo sin truncar
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
pd.set_option('display.width', None)
pd.set_option('display.max_colwidth', None)

max_idx = len(df_total) - 1

# 1. Creamos la matriz con todas las listas que tenías comentadas
positions_matrix = [
    [2026, 10231, 123123],
    [401234, 123123, 1500123],                                                                                                                                                                                                                                                                                                                                                                                                               
    [712039, 123123, 98234],
    [95736, 12314, 103982],
    [190385, 19374, 10329]
]

# 2. Abrimos el archivo UNA SOLA VEZ para todo el proceso
with open('convinaciones_de_queries.log', 'a', encoding='utf-8') as log_file:
    log_file.write(f"Total de filas en el dataframe: {len(df_total)}\n\n")
    
    # 3. Recorremos la matriz. 'i' es el índice (0, 1, 2...) y 'queue' es la lista actual
    for i, queue in enumerate(positions_matrix):
        NUMERO = i + 1 # Para que empiece en Query 1, Query 2, etc.
        
        # Filtrar solo las posiciones que existen PARA ESTA LISTA (queue)
        valid_positions = [p for p in queue if p <= max_idx]
        
        # Encabezados de la Query actual (en archivo y terminal)
        log_file.write(separator + "\n")
        log_file.write(f"Query {NUMERO}\n")
        log_file.write(separator + "\n")
        
        print(separator)
        print(f"Query {NUMERO}")
        print(separator)
        
        # Procesamos las posiciones de esta lista
        if valid_positions:
            for pos in valid_positions:
                header = f"FILA EN POSICIÓN {pos}:"
                
                print(header)
                log_file.write(header + "\n")
                log_file.write("----" * 20 + "\n")
                
                row = df_total.iloc[pos]
                row_str = row.to_string()
                
                print(row_str)
                print("\n")
                log_file.write(row_str + "\n\n")
        else:
            msg = f"⚠️ Ninguna de las posiciones {queue} es válida para un dataframe con {len(df_total)} filas"
            print(msg + "\n")
            log_file.write(msg + "\n\n")

print(f"\n✅ Resultados de las {len(positions_matrix)} queries guardados en 'convinaciones_de_queries.log' (modo append)")
```
